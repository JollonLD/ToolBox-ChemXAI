```python
# Função corrigida
def _rank_explanation(explanation):
    return np.argsort(explanation)[::-1].copy()

# Dados de teste simples
data = torch.tensor([
    [1.5, 2.3, 0.8, 1.2, 3.1],
    [2.1, 1.7, 1.5, 0.9, 2.8]
], dtype=torch.float64)

explanation = np.array([0.1, 0.8, 0.3, 0.05, 0.9])
n_features = 3

# Teste
mask = torch.zeros(data.shape, dtype=torch.float64)
important_features = _rank_explanation(explanation)[:n_features]
mask[:, important_features] = data[:, important_features]

print("Data:", data)
print("Explanation:", explanation)
print("Important features:", important_features)
print("Mask:", mask)
```

```python
for data in mask:
    print(data)
```
