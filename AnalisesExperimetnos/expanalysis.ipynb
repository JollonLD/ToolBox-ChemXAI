```python
import pandas as pd
import matplotlib.pyplot as plt
import json
import os
from pathlib import Path
```

```python
explanations_shap = []
explanations_lime = []

experiments_path = Path('experiment_explanations')

for pasta_filha in experiments_path.iterdir():
    if pasta_filha.is_dir():
        # Buscar arquivos LIME
        lime_file = pasta_filha / 'lime_results.json'
        if lime_file.exists():
            with open(lime_file, 'r', encoding='utf-8') as f:
                lime_data = json.load(f)
                explanations_lime.append({
                    'experiment': pasta_filha.name,
                    'data': lime_data
                })
        
        shap_file = pasta_filha / 'shap_results.json'
        if shap_file.exists():
            with open(shap_file, 'r', encoding='utf-8') as f:
                shap_data = json.load(f)
                explanations_shap.append({
                    'experiment': pasta_filha.name,
                    'data': shap_data
                })

print(f"Encontrados {len(explanations_lime)} arquivos LIME")
print(f"Encontrados {len(explanations_shap)} arquivos SHAP")
```

```python
print(explanations_shap[0]['data'])
```

```python
list_desc_lime = [exp['data']['model_type'].split('_')[0] for exp in explanations_lime]

properties = [
            'Rotational constant A: GHz', 'Rotational constant B: GHz', 'Rotational constant C: GHz',
            'Dipole moment (μ): Debye (D)', 'Isotropic polarizability (α): atomic units (a.u.)',
            'Energy of HOMO (ϵHOMO): Hartree (Ha)', 'Energy of LUMO (ϵLUMO): Hartree (Ha)',
            'Gap (ϵgap): Hartree (Ha)', 'Electronic spatial extent: atomic units (a.u.)',
            'Zero point vibrational energy (zpve): Hartree (Ha)', 'Internal energy at 0 K (U0): Hartree (Ha)',
            'Internal energy at 298.15 K (U): Hartree (Ha)', 'Enthalpy at 298.15 K (H): Hartree (Ha)',
            'Free energy at 298.15 K (G): Hartree (Ha)', 'Heat capacity at 298.15 K (Cv): cal/mol·K'
            ]

prop_lime = []
for exp in explanations_lime:
    model_type = exp['data']['model_type']
    parts = model_type.split('_')
    att_part = parts[1]
    index = int(att_part[3:])
    prop_lime.append(properties[index])


descriptors_lime = {
    'descriptor': list_desc_lime,
    'predicted_property': prop_lime,
    'feature_names': [exp['data']['feature_names'] for exp in explanations_lime]
}
```

```python
df_descriptors_lime = pd.DataFrame(descriptors_lime)

df_descriptors_lime
```

```python
list_desc_shap = [exp['data']['model_type'].split('_')[0] for exp in explanations_shap]

prop_shap = []
for exp in explanations_shap:
    model_type = exp['data']['model_type']
    parts = model_type.split('_')
    att_part = parts[1]
    index = int(att_part[3:])
    prop_shap.append(properties[index])


descriptors_shap = {
    'descriptor': list_desc_shap,
    'predicted_property': prop_shap,
    'feature_names': [exp['data']['feature_names'] for exp in explanations_shap]
}
```

```python
df_descriptors_shap = pd.DataFrame(descriptors_shap)

df_descriptors_shap
```

```python
top1_lime = []
top2_lime = []
top3_lime = []
top4_lime = []
top5_lime = []

features_lime = []

for exp in explanations_lime:
    local_exp = exp['data']['local_explanation']
    features_lime = exp['data']['feature_names'] 

    indexed_exp = [(i, val) for i, val in enumerate(local_exp)]
    
    sorted_exp = sorted(indexed_exp, key=lambda x: abs(x[1]), reverse=True)
    
    top1_lime.append(features_lime[sorted_exp[0][0]] if len(sorted_exp) > 0 else None)
    top2_lime.append(features_lime[sorted_exp[1][0]] if len(sorted_exp) > 1 else None)
    top3_lime.append(features_lime[sorted_exp[2][0]] if len(sorted_exp) > 2 else None)
    top4_lime.append(features_lime[sorted_exp[3][0]] if len(sorted_exp) > 3 else None)
    top5_lime.append(features_lime[sorted_exp[4][0]] if len(sorted_exp) > 4 else None)

top1_shap = []
top2_shap = []
top3_shap = []
top4_shap = []
top5_shap = []

features_shap = []

for exp in explanations_shap:
    local_exp = exp['data']['local_explanation']
    features_shap = exp['data']['feature_names'] 

    indexed_exp = [(i, val) for i, val in enumerate(local_exp)]
    
    sorted_exp = sorted(indexed_exp, key=lambda x: abs(x[1]), reverse=True)
    
    top1_shap.append(features_shap[sorted_exp[0][0]] if len(sorted_exp) > 0 else None)
    top2_shap.append(features_shap[sorted_exp[1][0]] if len(sorted_exp) > 1 else None)
    top3_shap.append(features_shap[sorted_exp[2][0]] if len(sorted_exp) > 2 else None)
    top4_shap.append(features_shap[sorted_exp[3][0]] if len(sorted_exp) > 3 else None)
    top5_shap.append(features_shap[sorted_exp[4][0]] if len(sorted_exp) > 4 else None)

explanation_lime_ranking = {
    'descriptor': list_desc_lime,
    'predicted_property': prop_lime,
    'top1': top1_lime,
    'top2': top2_lime,
    'top3': top3_lime,
    'top4': top4_lime,
    'top5': top5_lime
}

explanation_shap_ranking = {
    'descriptor': list_desc_shap,
    'predicted_property': prop_shap,
    'top1': top1_shap,
    'top2': top2_shap,
    'top3': top3_shap,
    'top4': top4_shap,
    'top5': top5_shap
}

df_top_exp_lime = pd.DataFrame(explanation_lime_ranking)
df_top_exp_shap = pd.DataFrame(explanation_shap_ranking)
```
